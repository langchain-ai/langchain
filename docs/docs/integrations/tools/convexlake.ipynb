### Credentials

Get a free API key at [convexlake.com](https://convexlake.com) and set it as an environment variable:

```python
import getpass
import os

if not os.environ.get("CONVEXLAKE_API_KEY"):
    os.environ["CONVEXLAKE_API_KEY"] = getpass.getpass("ConvexLake API key: ")
```

## Instantiation

```python
from langchain_convexlake import get_tools

tools = get_tools()  # reads CONVEXLAKE_API_KEY from environment
print([t.name for t in tools])
```

## Invocation

### Browse the catalog

```python
from langchain_convexlake import ConvexLakeCatalog

catalog = ConvexLakeCatalog()

# List all exchanges
print(catalog.invoke({}))

# Drill into Kalshi
print(catalog.invoke({"exchange": "kalshi"}))
```

### Fetch trade history

```python
from langchain_convexlake import ConvexLakeTrades

trades = ConvexLakeTrades()

# .func() returns (summary_str, full_dataframe)
summary, df = trades.func(
    exchange="kalshi",
    ticker="crypto",
    date="2026-10-01",
    slug="KXBTC15M-26OCT01-trades.csv",
)
print(summary)
df.head()
```

### Leaderboard

```python
from langchain_convexlake import ConvexLakeLeaderboard

lb = ConvexLakeLeaderboard()
summary, df = lb.func(exchange="polymarket", date="2026-10-01")
print(summary)
```

## Use within an agent

```python
from langchain.agents import create_agent
from langchain_convexlake import get_tools

tools = get_tools()
agent = create_agent(model="anthropic:claude-sonnet-5-5", tools=tools)

result = agent.invoke({
    "messages": [{
        "role": "user",
        "content": "Who are the top Polymarket traders from last week? Show their total P&L.",
    }]
})
print(result["messages"][-1].content)
```

### Accessing the DataFrame artifact

Tools that return tabular data use `response_format="content_and_artifact"`. In an agent run, the full DataFrame is available on `ToolMessage.artifact`:

```python
from langchain_core.messages import ToolMessage

for msg in result["messages"]:
    if isinstance(msg, ToolMessage) and msg.artifact is not None:
        print(type(msg.artifact))  # pandas.DataFrame
        print(msg.artifact.shape)
```
